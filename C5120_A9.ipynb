```python
import sqlite3

conn = sqlite3.connect('companydb.sqlite')
cursor = conn.cursor()

cursor.execute("""
    SELECT name
    FROM sqlite_master
    WHERE type = 'trigger'
""")

print("Triggers:", cursor.fetchall())
```

```python
# Insert a new person
cursor.execute("""
    INSERT INTO person (person_name, experience)
    VALUES ('Colab Test Person', 3)
""")

person_id = cursor.lastrowid
conn.commit()

print("New person inserted with ID:", person_id)

# Verify INSERT trigger fired
print("\nAfter INSERT:")
cursor.execute(
    "SELECT * FROM person_log WHERE person_id = ?",
    (person_id,)
)
for row in cursor.fetchall():
    print(row)

# Delete the test person
cursor.execute(
    "DELETE FROM person WHERE person_id = ?",
    (person_id,)
)
conn.commit()

# Verify DELETE trigger fired
print("\nAfter DELETE:")
cursor.execute(
    "SELECT * FROM person_log WHERE person_id = ? ORDER BY log_id",
    (person_id,)
)
for row in cursor.fetchall():
    print(row)

# Confirm person was removed
cursor.execute(
    "SELECT * FROM person WHERE person_id = ?",
    (person_id,)
)
print("\nPerson remaining in database:", cursor.fetchall())

conn.close()
```
